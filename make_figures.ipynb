```python
from PIL import Image, ImageDraw, ImageFont
import os

def create_figure(top_img_path, bottom_left_path, bottom_right_path, output_path, dpi=300):
    """
    Create a publication figure with one image on top (A) and two on bottom (B, C).
    
    Args:
        top_img_path: Path to top image (A)
        bottom_left_path: Path to bottom left image (B)
        bottom_right_path: Path to bottom right image (C)
        output_path: Path for output figure
        dpi: DPI for output image
    """
    # Load images
    img_a = Image.open(top_img_path)
    img_b = Image.open(bottom_left_path)
    img_c = Image.open(bottom_right_path)
    
    # Calculate dimensions
    padding = 40  # Padding between images and edges
    label_space = 280  # Space for labels above images (increased for 255pt font)
    
    # Canvas width based on top image
    canvas_width = img_a.width + 2 * padding
    
    # Bottom images get 50/50 of available space
    bottom_img_width = (canvas_width - 3 * padding) // 2
    
    # Resize bottom images to fit 50/50 space while maintaining aspect ratio
    b_ratio = img_b.height / img_b.width
    c_ratio = img_c.height / img_c.width
    
    img_b_resized = img_b.resize((bottom_img_width, int(bottom_img_width * b_ratio)), Image.LANCZOS)
    img_c_resized = img_c.resize((bottom_img_width, int(bottom_img_width * c_ratio)), Image.LANCZOS)
    
    # Height: top image + tallest bottom image + spacing + labels
    bottom_height = max(img_b_resized.height, img_c_resized.height)
    canvas_height = (img_a.height + bottom_height + 3 * padding + 3 * label_space)
    
    # Create canvas
    canvas = Image.new('RGB', (canvas_width, canvas_height), 'white')
    draw = ImageDraw.Draw(canvas)
    
    # Try to use a nice font, fall back to default if not available
    try:
        font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", 255)
    except:
        font = ImageFont.load_default()
    
    # Position and paste image A (top, centered)
    x_a = (canvas_width - img_a.width) // 2
    y_a = padding + label_space
    canvas.paste(img_a, (x_a, y_a))
    draw.text((x_a, padding), "A)", fill='black', font=font)
    
    # Position and paste image B (bottom left, 50% width)
    x_b = padding
    y_b = y_a + img_a.height + padding + label_space * 2
    canvas.paste(img_b_resized, (x_b, y_b))
    
    # Position and paste image C (bottom right, 50% width, aligned with B)
    x_c = x_b + bottom_img_width + padding
    y_c = y_b - label_space * 2  # Raised relative to B
    canvas.paste(img_c_resized, (x_c, y_c))
    
    # Draw labels at the same height (above the higher image C)
    label_y = y_c - label_space + 10
    draw.text((x_b, label_y), "B)", fill='black', font=font)
    draw.text((x_c, label_y), "C)", fill='black', font=font)
    
    # Save with high quality
    canvas.save(output_path, dpi=(dpi, dpi), quality=95)
    print(f"Figure saved to {output_path}")
    print(f"Dimensions: {canvas_width}x{canvas_height} pixels")

# Example usage
if __name__ == "__main__":
    create_figure(
        top_img_path="WVS_v6/language_families_map.png",
        bottom_left_path="WVS_v6/language_families_pie_chart.png",
        bottom_right_path="WVS_v6/questions_categories_pie_chart.png",
        output_path="figure1_overview.png",
        dpi=300
    )
```

```python
from PIL import Image, ImageDraw, ImageFont
import os

def create_figure2(top_left_path, top_right_path, bottom_left_path, bottom_right_path, output_path, dpi=300):
    """
    Create a publication figure with 2x2 grid layout (A, B on top; C, D on bottom).
    
    Args:
        top_left_path: Path to top left image (A)
        top_right_path: Path to top right image (B)
        bottom_left_path: Path to bottom left image (C)
        bottom_right_path: Path to bottom right image (D)
        output_path: Path for output figure
        dpi: DPI for output image
    """
    # Load images
    img_a = Image.open(top_left_path)
    img_b = Image.open(top_right_path)
    img_c = Image.open(bottom_left_path)
    img_d = Image.open(bottom_right_path)
    
    # Calculate dimensions
    padding = 40  # Padding between images and edges
    label_space = 280  # Space for labels above images
    
    # Determine canvas width from the wider row
    top_width = img_a.width + img_b.width + 3 * padding
    bottom_width = img_c.width + img_d.width + 3 * padding
    canvas_width = max(top_width, bottom_width)
    
    # Each column gets 50% of available space
    col_width = (canvas_width - 3 * padding) // 2
    
    # Resize all images to fit 50/50 space while maintaining aspect ratio
    a_ratio = img_a.height / img_a.width
    b_ratio = img_b.height / img_b.width
    c_ratio = img_c.height / img_c.width
    d_ratio = img_d.height / img_d.width
    
    img_a_resized = img_a.resize((col_width, int(col_width * a_ratio)), Image.LANCZOS)
    img_b_resized = img_b.resize((col_width, int(col_width * b_ratio)), Image.LANCZOS)
    img_c_resized = img_c.resize((col_width, int(col_width * c_ratio)), Image.LANCZOS)
    img_d_resized = img_d.resize((col_width, int(col_width * d_ratio)), Image.LANCZOS)
    
    # Calculate heights for each row
    top_row_height = max(img_a_resized.height, img_b_resized.height)
    bottom_row_height = max(img_c_resized.height, img_d_resized.height)
    
    # Total canvas height
    canvas_height = (top_row_height + bottom_row_height + 
                     4 * padding + 4 * label_space)
    
    # Create canvas
    canvas = Image.new('RGB', (canvas_width, canvas_height), 'white')
    draw = ImageDraw.Draw(canvas)
    
    # Try to use a nice font, fall back to default if not available
    try:
        font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", 255)
    except:
        font = ImageFont.load_default()
    
    # Position and paste top row images (A and B)
    x_a = padding
    y_a = padding + label_space
    canvas.paste(img_a_resized, (x_a, y_a))
    
    x_b = x_a + col_width + padding
    y_b = y_a + 2*label_space  # Lower B but keep its label aligned with A
    canvas.paste(img_b_resized, (x_b, y_b))
    
    # Position and paste bottom row images (C and D)
    x_c = padding
    y_c = y_a + top_row_height + padding + label_space
    canvas.paste(img_c_resized, (x_c, y_c))
    
    x_d = x_c + col_width + padding
    y_d = y_c
    canvas.paste(img_d_resized, (x_d, y_d))
    
    # Draw labels - aligned within each row
    top_label_y = padding
    draw.text((x_a, top_label_y), "A)", fill='black', font=font)
    draw.text((x_b, top_label_y), "B)", fill='black', font=font)
    
    bottom_label_y = y_c - label_space + 10
    draw.text((x_c, bottom_label_y), "C)", fill='black', font=font)
    draw.text((x_d, bottom_label_y), "D)", fill='black', font=font)
    
    # Save with high quality
    canvas.save(output_path, dpi=(dpi, dpi), quality=95)
    print(f"Figure saved to {output_path}")
    print(f"Dimensions: {canvas_width}x{canvas_height} pixels")

# Example usage
if __name__ == "__main__":
    create_figure2(
        top_left_path="WVS_v6/filtered_clustermaplang_imp.png",
        top_right_path="WVS_v6/filtered_pca_pc1_pc2_imp.png",
        bottom_left_path="WVS_v6/filtered_biplotPCA_circ_Religion.png",
        bottom_right_path="WVS_v6/filtered_biplotPCA_circ_Politics.png",
        output_path="figure2_clusters.png",
        dpi=300
    )
```

```python
from PIL import Image, ImageDraw, ImageFont
import os

def create_figure3(left_path, middle_path, right_path, output_path, dpi=300):
    """
    Create a publication figure with three images side by side (A, B, C).
    
    Args:
        left_path: Path to left image (A)
        middle_path: Path to middle image (B)
        right_path: Path to right image (C)
        output_path: Path for output figure
        dpi: DPI for output image
    """
    # Load images
    img_a = Image.open(left_path)
    img_b = Image.open(middle_path)
    img_c = Image.open(right_path)
    
    # Calculate dimensions
    padding = 40  # Padding between images and edges
    label_space = 280  # Space for labels above images
    
    # Calculate canvas width - all three images get equal space
    total_width = img_a.width + img_b.width + img_c.width + 4 * padding
    canvas_width = total_width
    
    # Each column gets 1/3 of available space
    col_width = (canvas_width - 4 * padding) // 3
    
    # Resize all images to fit 1/3 space while maintaining aspect ratio
    a_ratio = img_a.height / img_a.width
    b_ratio = img_b.height / img_b.width
    c_ratio = img_c.height / img_c.width
    
    img_a_resized = img_a.resize((col_width, int(col_width * a_ratio)), Image.LANCZOS)
    img_b_resized = img_b.resize((col_width, int(col_width * b_ratio)), Image.LANCZOS)
    img_c_resized = img_c.resize((col_width, int(col_width * c_ratio)), Image.LANCZOS)
    
    # Calculate canvas height
    max_height = max(img_a_resized.height, img_b_resized.height, img_c_resized.height)
    canvas_height = max_height + 2 * padding + label_space
    
    # Create canvas
    canvas = Image.new('RGB', (canvas_width, canvas_height), 'white')
    draw = ImageDraw.Draw(canvas)
    
    # Try to use a nice font, fall back to default if not available
    try:
        font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", 255)
    except:
        font = ImageFont.load_default()
    
    # Position and paste images side by side
    x_a = padding
    y_a = padding + label_space
    canvas.paste(img_a_resized, (x_a, y_a))
    
    x_b = x_a + col_width + padding
    y_b = y_a
    canvas.paste(img_b_resized, (x_b, y_b))
    
    x_c = x_b + col_width + padding
    y_c = y_a
    canvas.paste(img_c_resized, (x_c, y_c))
    
    # Draw labels - all aligned at the same height
    label_y = padding
    draw.text((x_a, label_y), "A)", fill='black', font=font)
    draw.text((x_b, label_y), "B)", fill='black', font=font)
    draw.text((x_c, label_y), "C)", fill='black', font=font)
    
    # Save with high quality
    canvas.save(output_path, dpi=(dpi, dpi), quality=95)
    print(f"Figure saved to {output_path}")
    print(f"Dimensions: {canvas_width}x{canvas_height} pixels")

# Example usage
if __name__ == "__main__":
    create_figure3(
        left_path="WVS_v6/filtered_nmf_c2_imp.png",
        middle_path="WVS_v6/filtered_nmf_c3_imp.png",
        right_path="WVS_v6/filtered_nmf_c3_triangle.png",
        output_path="figure3_nmf.png",
        dpi=300
    )
```

```python
from PIL import Image, ImageDraw, ImageFont
import os

def create_figure6(row1_path, row2_left_path, row2_right_path, row3_left_path, row3_right_path, output_path, dpi=300):
    """
    Create a publication figure with 3 rows:
    - Row 1: Single image A (centered)
    - Row 2: Two images B and C (side by side)
    - Row 3: Two images D and E (side by side)
    
    Args:
        row1_path: Path to row 1 image (A)
        row2_left_path: Path to row 2 left image (B)
        row2_right_path: Path to row 2 right image (C)
        row3_left_path: Path to row 3 left image (D)
        row3_right_path: Path to row 3 right image (E)
        output_path: Path for output figure
        dpi: DPI for output image
    """
    # Load images
    img_a = Image.open(row1_path)
    img_b = Image.open(row2_left_path)
    img_c = Image.open(row2_right_path)
    img_d = Image.open(row3_left_path)
    img_e = Image.open(row3_right_path)
    
    # Calculate dimensions
    padding = 40  # Padding between images and edges
    label_space = 280  # Space for labels above images
    
    # Canvas width based on row 1 image
    canvas_width = img_a.width + 2 * padding
    
    # Each column in rows 2 and 3 gets 50% of available space
    col_width = (canvas_width - 3 * padding) // 2
    
    # Resize images while maintaining aspect ratio
    # Row 1: keep original size
    img_a_resized = img_a
    
    # Row 2: resize to 50/50
    b_ratio = img_b.height / img_b.width
    c_ratio = img_c.height / img_c.width
    img_b_resized = img_b.resize((col_width, int(col_width * b_ratio)), Image.LANCZOS)
    img_c_resized = img_c.resize((col_width, int(col_width * c_ratio)), Image.LANCZOS)
    
    # Row 3: resize to 50/50
    d_ratio = img_d.height / img_d.width
    e_ratio = img_e.height / img_e.width
    img_d_resized = img_d.resize((col_width, int(col_width * d_ratio)), Image.LANCZOS)
    img_e_resized = img_e.resize((col_width, int(col_width * e_ratio)), Image.LANCZOS)
    
    # Calculate heights for each row
    row1_height = img_a_resized.height
    row2_height = max(img_b_resized.height, img_c_resized.height)
    row3_height = max(img_d_resized.height, img_e_resized.height)
    
    # Total canvas height
    canvas_height = (row1_height + row2_height + row3_height + 
                     5 * padding + 5 * label_space)
    
    # Create canvas
    canvas = Image.new('RGB', (canvas_width, canvas_height), 'white')
    draw = ImageDraw.Draw(canvas)
    
    # Try to use a nice font, fall back to default if not available
    try:
        font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", 255)
    except:
        font = ImageFont.load_default()
    
    # Row 1: Position and paste image A (full width)
    x_a = padding
    y_a = padding + label_space
    canvas.paste(img_a_resized, (x_a, y_a))
    draw.text((x_a, padding), "A)", fill='black', font=font)
    
    # Row 2: Position and paste images B and C
    x_b = padding
    y_b = y_a + row1_height + padding + label_space
    canvas.paste(img_b_resized, (x_b, y_b))
    
    x_c = x_b + col_width + padding
    y_c = y_b
    canvas.paste(img_c_resized, (x_c, y_c))
    
    row2_label_y = y_b - label_space + 10
    draw.text((x_b, row2_label_y), "B)", fill='black', font=font)
    draw.text((x_c, row2_label_y), "C)", fill='black', font=font)
    
    # Row 3: Position and paste images D and E
    x_d = padding
    y_d = y_b + row2_height + padding + label_space
    canvas.paste(img_d_resized, (x_d, y_d))
    
    x_e = x_d + col_width + padding
    y_e = y_d
    canvas.paste(img_e_resized, (x_e, y_e))
    
    row3_label_y = y_d - label_space + 10
    draw.text((x_d, row3_label_y), "D)", fill='black', font=font)
    draw.text((x_e, row3_label_y), "E)", fill='black', font=font)
    
    # Save with high quality
    canvas.save(output_path, dpi=(dpi, dpi), quality=95)
    print(f"Figure saved to {output_path}")
    print(f"Dimensions: {canvas_width}x{canvas_height} pixels")

# Example usage
if __name__ == "__main__":
    create_figure6(
        row1_path="WVS_v6/USAPolitics_filter_pca_pc1_pc2_imp.png",
        row2_left_path="WVS_v6/USAPolitics_filter_distances_sorted_USD.png",
        row2_right_path="WVS_v6/USAPolitics_filter_distances_sorted_USR.png",
        row3_left_path="WVS_v6/16-29_CADUSA_distances_sorted_CDE.png",
        row3_right_path="WVS_v6/50andOver_CADUSA_distances_sorted_CDE.png",
        output_path="figure6_usa_politics.png",
        dpi=300
    )
```
